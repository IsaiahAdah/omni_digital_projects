```python
import csv
import random
```

```python
# total module
female = 8000
male = 7000
```

```python
def get_age_group(age):
    if age <= 12:
        return "Child"
    elif age <= 19:
        return "Teen"
    elif age <= 59:
        return "Adult"
    else:
        return "Senior"

def get_diet(age, gender):
    if age <= 12:
        return "Growth Diet"
    elif age <= 19:
        return "High Energy Diet"
    elif age <= 59:
        if gender == "Male":
            return "High Protein Diet"
        else:
            return "Balanced Diet"
    else:
        return "Low Sodium Diet"
```

```python
users = []
```

```python
for _ in range(female):
    age       = random.randint(1, 100)
    gender    = "Female"
    age_group = get_age_group(age)
    diet      = get_diet(age, gender)
    users.append([age, age_group, gender, diet])

for _ in range(male):
    age       = random.randint(1, 100)
    gender    = "Male"
    age_group = get_age_group(age)
    diet      = get_diet(age, gender)
    users.append([age, age_group, gender, diet])

random.shuffle(users)
```

```python
with open("ai_doctor_users.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Age", "Age_Group", "Gender", "Diet"])
    writer.writerows(users)

print("Done! File saved as ai_doctor_users.csv")
print(f"Total records: {len(users)}")
```
